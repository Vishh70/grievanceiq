```python
# @title 🧠 Step 5/12 — Reload Datasets + MiniLM on T4 GPU

import os
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer

# ============================================================
# 1. GOOGLE DRIVE PATH
# ============================================================

DATA_DIR = "/content/drive/MyDrive/GrievanceIQ Dataset"

complaints_path = os.path.join(
    DATA_DIR,
    "grievanceiq_complaints_multilabel_6000.csv"
)

relationships_path = os.path.join(
    DATA_DIR,
    "grievanceiq_relationship_pairs_16000.csv"
)

# ============================================================
# 2. LOAD DATASETS
# ============================================================

complaints_df = pd.read_csv(complaints_path)
relationships_df = pd.read_csv(relationships_path)

print("===== GrievanceIQ Training Progress =====")
print("Step: 5 / 12")
print()

print("✅ Complaints loaded:", complaints_df.shape)
print("✅ Relationship pairs loaded:", relationships_df.shape)

# Expected:
# (6000, 20)
# (16000, 17)

# ============================================================
# 3. CREATE DRAINAGE FLAG
# ============================================================

complaints_df["drainage_flag"] = (
    complaints_df["issue_types"]
    .fillna("")
    .astype(str)
    .str.split("|")
    .apply(
        lambda labels: int(
            "Drainage" in [x.strip() for x in labels]
        )
    )
)

# ============================================================
# 4. DEFINE 9 TARGET LABELS
# ============================================================

issue_flag_columns = [
    "road_damage_flag",
    "roadside_flooding_flag",
    "water_leakage_flag",
    "electric_pole_flag",
    "streetlight_flag",
    "traffic_signal_flag",
    "garbage_flag",
    "tree_hazard_flag",
    "drainage_flag"
]

print("\n✅ Number of issue labels:", len(issue_flag_columns))

# ============================================================
# 5. CHECK GPU
# ============================================================

print("\n--- GPU ---")

if torch.cuda.is_available():
    device = "cuda"
    print("✅ GPU:", torch.cuda.get_device_name(0))
else:
    device = "cpu"
    print("⚠️ GPU unavailable, using CPU")

# ============================================================
# 6. LOAD PRETRAINED MINILM
# ============================================================

print("\n--- Loading MiniLM ---")
print("Model: sentence-transformers/all-MiniLM-L6-v2")

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device=device
)

embedding_dimension = embedding_model.get_embedding_dimension()

print("\n✅ MiniLM loaded successfully")
print("Device:", device)
print("Embedding dimension:", embedding_dimension)

# ============================================================
# 7. FINAL VERIFICATION
# ============================================================

print("\n===== Verification =====")

print("Complaints:", len(complaints_df))
print("Relationship pairs:", len(relationships_df))
print("Issue labels:", len(issue_flag_columns))
print("Embedding dimension:", embedding_dimension)

if device == "cuda" and embedding_dimension == 384:
    print("\n🎉 STEP 5 COMPLETE")
    print("Ready for Step 6: Generate 6,000 complaint embeddings")
else:
    print("\n⚠️ Please check GPU/model configuration")
```

```python
# @title 🔢 Step 6/12 — Generate 6,000 MiniLM Complaint Embeddings

import os
import time
import numpy as np
from tqdm.auto import tqdm

# ============================================================
# CONFIGURATION
# ============================================================

BATCH_SIZE = 64

embedding_save_path = os.path.join(
    DATA_DIR,
    "complaint_embeddings_6000.npy"
)

# ============================================================
# PREPARE TEXT
# ============================================================

complaint_texts = (
    complaints_df["complaint_text"]
    .fillna("")
    .astype(str)
    .tolist()
)

TOTAL_COMPLAINTS = len(complaint_texts)

print("===== GrievanceIQ Training Progress =====")
print("Step: 6 / 12")
print("Task: Generate MiniLM embeddings")
print()
print("Total complaints:", TOTAL_COMPLAINTS)
print("Batch size:", BATCH_SIZE)
print()

# ============================================================
# CHECK WHETHER EMBEDDINGS ALREADY EXIST
# ============================================================

if os.path.exists(embedding_save_path):

    print("⚠️ Existing embedding file found:")
    print(embedding_save_path)

    existing_embeddings = np.load(embedding_save_path)

    print("Existing shape:", existing_embeddings.shape)

    if existing_embeddings.shape == (
        TOTAL_COMPLAINTS,
        embedding_model.get_embedding_dimension()
    ):
        print("\n✅ Existing embeddings are valid.")
        print("Skipping generation.")
        complaint_embeddings = existing_embeddings

    else:
        print("\n⚠️ Existing file has unexpected shape.")
        print("Regenerating embeddings...")
        os.remove(embedding_save_path)

# ============================================================
# GENERATE EMBEDDINGS IF NEEDED
# ============================================================

if "complaint_embeddings" not in locals():

    total_batches = (
        TOTAL_COMPLAINTS + BATCH_SIZE - 1
    ) // BATCH_SIZE

    print("Total batches:", total_batches)
    print()
    print("Starting embedding generation...\n")

    start_time = time.time()

    all_embeddings = []

    progress_bar = tqdm(
        range(0, TOTAL_COMPLAINTS, BATCH_SIZE),
        total=total_batches,
        desc="Embedding batches",
        unit="batch"
    )

    for batch_start in progress_bar:

        batch_end = min(
            batch_start + BATCH_SIZE,
            TOTAL_COMPLAINTS
        )

        batch_texts = complaint_texts[batch_start:batch_end]

        batch_embeddings = embedding_model.encode(
            batch_texts,
            batch_size=BATCH_SIZE,
            show_progress_bar=False,
            normalize_embeddings=True,
            convert_to_numpy=True
        )

        all_embeddings.append(batch_embeddings)

        completed = batch_end
        remaining = TOTAL_COMPLAINTS - completed

        elapsed = time.time() - start_time

        progress_bar.set_postfix(
            completed=f"{completed}/{TOTAL_COMPLAINTS}",
            remaining=remaining,
            elapsed=f"{elapsed/60:.1f} min"
        )

    # Combine all batches
    complaint_embeddings = np.vstack(all_embeddings)

    # ========================================================
    # SAVE
    # ========================================================

    np.save(
        embedding_save_path,
        complaint_embeddings
    )

    elapsed = time.time() - start_time

    print("\n✅ Embedding generation complete")
    print("Time taken:", round(elapsed / 60, 2), "minutes")

# ============================================================
# FINAL VERIFICATION
# ============================================================

print("\n===== STEP 6 VERIFICATION =====")

print("Embedding shape:", complaint_embeddings.shape)
print(
    "Expected shape:",
    (
        TOTAL_COMPLAINTS,
        embedding_model.get_embedding_dimension()
    )
)

print("Saved to:")
print(embedding_save_path)

# Check numeric values
print("\nFirst embedding:")
print(complaint_embeddings[0][:10])

print("\nData type:")
print(complaint_embeddings.dtype)

# Final checks
assert complaint_embeddings.shape == (
    TOTAL_COMPLAINTS,
    embedding_model.get_embedding_dimension()
)

assert np.isfinite(complaint_embeddings).all()

print("\n✅ Shape check passed")
print("✅ NaN/Inf check passed")
print("✅ 6,000 complaint embeddings ready")

print("\n🎉 STEP 6 COMPLETE")
print("Next: Step 7/12 — Grouped Train / Validation / Test Split")
```

```python
# @title ✂️ Step 7/12 — Grouped Train / Validation / Test Split

from sklearn.model_selection import GroupShuffleSplit
import numpy as np

print("===== GrievanceIQ Training Progress =====")
print("Step: 7 / 12")
print()

# ============================================================
# 1. PREPARE FEATURES AND TARGETS
# ============================================================

X = complaint_embeddings

Y = complaints_df[issue_flag_columns].astype(int).values

groups = complaints_df["issue_id"].astype(str).values

print("Total complaints:", len(X))
print("Embedding shape:", X.shape)
print("Target shape:", Y.shape)
print("Unique issue groups:", len(np.unique(groups)))

# ============================================================
# 2. FIRST SPLIT
#    85% train+validation
#    15% test
# ============================================================

gss_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

train_val_idx, test_idx = next(
    gss_test.split(
        X,
        Y,
        groups=groups
    )
)

# ============================================================
# 3. SECOND SPLIT
#    From the remaining 85%:
#    ~82.35% -> train
#    ~17.65% -> validation
#
# This gives approximately:
# train = 70% total
# val   = 15% total
# test  = 15% total
# ============================================================

train_val_groups = groups[train_val_idx]

gss_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.17647058823529413,
    random_state=42
)

train_rel_idx, val_rel_idx = next(
    gss_val.split(
        X[train_val_idx],
        Y[train_val_idx],
        groups=train_val_groups
    )
)

train_idx = train_val_idx[train_rel_idx]
val_idx = train_val_idx[val_rel_idx]

# ============================================================
# 4. DISPLAY COUNTS
# ============================================================

print("\n===== SPLIT COUNTS =====")

print("Total records:     ", len(X))
print("Training records:  ", len(train_idx))
print("Validation records:", len(val_idx))
print("Test records:      ", len(test_idx))

print("\n===== PERCENTAGES =====")

print(
    f"Training:   {len(train_idx)/len(X)*100:.2f}%"
)

print(
    f"Validation: {len(val_idx)/len(X)*100:.2f}%"
)

print(
    f"Test:       {len(test_idx)/len(X)*100:.2f}%"
)

# ============================================================
# 5. CHECK TOTAL
# ============================================================

total_after_split = (
    len(train_idx)
    + len(val_idx)
    + len(test_idx)
)

remaining = len(X) - total_after_split

print("\n===== RECORD ACCOUNTING =====")
print("Total original:", len(X))
print("Train + Validation + Test:", total_after_split)
print("Remaining:", remaining)

assert total_after_split == len(X)
assert remaining == 0

print("✅ All 6,000 records accounted for")

# ============================================================
# 6. CHECK FOR ISSUE_ID LEAKAGE
# ============================================================

train_groups = set(groups[train_idx])
val_groups = set(groups[val_idx])
test_groups = set(groups[test_idx])

train_val_overlap = train_groups.intersection(val_groups)
train_test_overlap = train_groups.intersection(test_groups)
val_test_overlap = val_groups.intersection(test_groups)

print("\n===== GROUP LEAKAGE CHECK =====")

print(
    "Train ↔ Validation overlapping issue_ids:",
    len(train_val_overlap)
)

print(
    "Train ↔ Test overlapping issue_ids:",
    len(train_test_overlap)
)

print(
    "Validation ↔ Test overlapping issue_ids:",
    len(val_test_overlap)
)

assert len(train_val_overlap) == 0
assert len(train_test_overlap) == 0
assert len(val_test_overlap) == 0

print("\n✅ No issue_id leakage detected")

# ============================================================
# 7. LABEL DISTRIBUTION
# ============================================================

print("\n===== LABEL DISTRIBUTION =====")

train_counts = complaints_df.iloc[train_idx][issue_flag_columns].sum()
val_counts = complaints_df.iloc[val_idx][issue_flag_columns].sum()
test_counts = complaints_df.iloc[test_idx][issue_flag_columns].sum()

distribution_df = np.column_stack([
    train_counts.values,
    val_counts.values,
    test_counts.values
])

import pandas as pd

distribution_table = pd.DataFrame(
    distribution_df,
    index=issue_flag_columns,
    columns=[
        "Train",
        "Validation",
        "Test"
    ]
)

display(distribution_table)

print("\n🎉 STEP 7 COMPLETE")
print("Next: Step 8/12 — Train the Multi-Label Classifier")
```

```python
# @title 🤖 Step 8/12 — Train Multi-Label Classifier

import os
import json
import time
import joblib
import numpy as np

from sklearn.linear_model import LogisticRegression

# ============================================================
# STEP INFORMATION
# ============================================================

print("===== GrievanceIQ Training Progress =====")
print("Step: 8 / 12")
print()

print("Training records:", len(train_idx))
print("Validation records:", len(val_idx))
print("Test records:", len(test_idx))
print("Number of labels:", len(issue_flag_columns))
print()

# ============================================================
# CHECK TRAINING DATA
# ============================================================

X_train = complaint_embeddings[train_idx]
Y_train = complaints_df.iloc[train_idx][issue_flag_columns].astype(int).values

print("Training embedding shape:", X_train.shape)
print("Training target shape:", Y_train.shape)

assert X_train.shape[0] == len(train_idx)
assert Y_train.shape[0] == len(train_idx)

# ============================================================
# CREATE MODEL DIRECTORY
# ============================================================

MODEL_DIR = os.path.join(
    DATA_DIR,
    "trained_models"
)

os.makedirs(MODEL_DIR, exist_ok=True)

CHECKPOINT_DIR = os.path.join(
    MODEL_DIR,
    "multilabel_checkpoints"
)

os.makedirs(CHECKPOINT_DIR, exist_ok=True)

print("\nModel directory:")
print(MODEL_DIR)

# ============================================================
# TRAIN 9 BINARY MODELS
# ============================================================

trained_models = {}

training_times = {}

TOTAL_LABELS = len(issue_flag_columns)

print("\n========================================")
print("STARTING MULTI-LABEL TRAINING")
print("========================================\n")

overall_start = time.time()

for label_number, label in enumerate(issue_flag_columns, start=1):

    remaining_labels = TOTAL_LABELS - label_number

    print("----------------------------------------")
    print(f"Label {label_number}/{TOTAL_LABELS}")
    print(f"Current label: {label}")
    print(f"Completed labels: {label_number - 1}")
    print(f"Remaining labels: {remaining_labels}")
    print("----------------------------------------")

    # --------------------------------------------------------
    # Get binary target
    # --------------------------------------------------------

    y_label = Y_train[:, label_number - 1]

    positive_count = int(y_label.sum())
    negative_count = int(len(y_label) - positive_count)

    print("Positive examples:", positive_count)
    print("Negative examples:", negative_count)

    # --------------------------------------------------------
    # Train Logistic Regression
    # --------------------------------------------------------

    model = LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        solver="liblinear",
        random_state=42
    )

    start_time = time.time()

    model.fit(
        X_train,
        y_label
    )

    elapsed = time.time() - start_time

    trained_models[label] = model
    training_times[label] = elapsed

    # --------------------------------------------------------
    # Save checkpoint immediately
    # --------------------------------------------------------

    checkpoint_path = os.path.join(
        CHECKPOINT_DIR,
        f"{label}.joblib"
    )

    joblib.dump(
        model,
        checkpoint_path
    )

    print(f"✅ Model trained for: {label}")
    print(f"Training time: {elapsed:.2f} seconds")
    print("Checkpoint saved:")
    print(checkpoint_path)

    print(
        f"Progress: {label_number}/{TOTAL_LABELS} "
        f"labels complete"
    )

# ============================================================
# SAVE ALL MODELS TO ONE FILE
# ============================================================

all_models_path = os.path.join(
    MODEL_DIR,
    "multilabel_classifier.joblib"
)

joblib.dump(
    trained_models,
    all_models_path
)

# ============================================================
# SAVE LABEL ORDER
# ============================================================

labels_path = os.path.join(
    MODEL_DIR,
    "issue_labels.json"
)

with open(labels_path, "w") as f:
    json.dump(
        issue_flag_columns,
        f,
        indent=2
    )

# ============================================================
# SAVE TRAINING CONFIGURATION
# ============================================================

config = {
    "embedding_model": "sentence-transformers/all-MiniLM-L6-v2",
    "embedding_dimension": 384,
    "classifier": "LogisticRegression",
    "solver": "liblinear",
    "max_iter": 3000,
    "class_weight": "balanced",
    "random_state": 42,
    "number_of_labels": len(issue_flag_columns),
    "labels": issue_flag_columns,
    "training_records": len(train_idx),
    "validation_records": len(val_idx),
    "test_records": len(test_idx),
    "group_split": True,
    "group_column": "issue_id"
}

config_path = os.path.join(
    MODEL_DIR,
    "multilabel_training_config.json"
)

with open(config_path, "w") as f:
    json.dump(
        config,
        f,
        indent=2
    )

# ============================================================
# FINAL SUMMARY
# ============================================================

overall_elapsed = time.time() - overall_start

print("\n========================================")
print("🎉 STEP 8 COMPLETE")
print("========================================")

print("\nLabels trained:", len(trained_models))
print("Expected labels:", TOTAL_LABELS)

print("\nTraining records used:", len(train_idx))
print("Validation records NOT used for fitting:", len(val_idx))
print("Test records NOT used for fitting:", len(test_idx))

print("\nTotal training time:")
print(f"{overall_elapsed / 60:.2f} minutes")

print("\n===== TRAINING TIME PER LABEL =====")

for label, seconds in training_times.items():
    print(f"{label}: {seconds:.2f} sec")

print("\n===== SAVED FILES =====")

print("Main model:")
print(all_models_path)

print("\nLabels:")
print(labels_path)

print("\nConfiguration:")
print(config_path)

print("\nCheckpoints:")
print(CHECKPOINT_DIR)

print("\nRemaining major steps: 4")
print("Next: Step 9/12 — Validation + threshold tuning + final test evaluation")
```

```python
# @title 📊 Step 9/12 — Validation + Threshold Tuning + Final Test Evaluation

import os
import json
import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    hamming_loss,
    accuracy_score,
    classification_report,
    multilabel_confusion_matrix
)

# ============================================================
# STEP INFORMATION
# ============================================================

print("===== GrievanceIQ Training Progress =====")
print("Step: 9 / 12")
print()

print("Training records:   ", len(train_idx))
print("Validation records: ", len(val_idx))
print("Test records:       ", len(test_idx))
print("Labels:              ", len(issue_flag_columns))

print("\nIMPORTANT:")
print("Training set = already used for fitting")
print("Validation set = used now for threshold tuning")
print("Test set = kept untouched until final evaluation")

# ============================================================
# MAKE SURE TRAINED MODELS EXIST
# ============================================================

if "trained_models" not in locals():

    print("\n⚠️ trained_models not found in memory.")
    print("Loading saved model from Google Drive...")

    all_models_path = os.path.join(
        DATA_DIR,
        "trained_models",
        "multilabel_classifier.joblib"
    )

    trained_models = joblib.load(all_models_path)

    print("✅ Saved model loaded")

# ============================================================
# PREPARE VALIDATION / TEST DATA
# ============================================================

X_val = complaint_embeddings[val_idx]
Y_val = complaints_df.iloc[val_idx][issue_flag_columns].astype(int).values

X_test = complaint_embeddings[test_idx]
Y_test = complaints_df.iloc[test_idx][issue_flag_columns].astype(int).values

print("\nValidation shape:")
print("X_val:", X_val.shape)
print("Y_val:", Y_val.shape)

print("\nTest shape:")
print("X_test:", X_test.shape)
print("Y_test:", Y_test.shape)

# ============================================================
# GET VALIDATION PROBABILITIES
# ============================================================

print("\n========================================")
print("GENERATING VALIDATION PROBABILITIES")
print("========================================")

val_probabilities = np.zeros(
    (len(val_idx), len(issue_flag_columns)),
    dtype=np.float32
)

for label_number, label in enumerate(issue_flag_columns, start=1):

    model = trained_models[label]

    val_probabilities[:, label_number - 1] = (
        model.predict_proba(X_val)[:, 1]
    )

    remaining = len(issue_flag_columns) - label_number

    print(
        f"Validation probability: "
        f"{label_number}/{len(issue_flag_columns)} | "
        f"Completed={label_number} | "
        f"Remaining={remaining}"
    )

print("\n✅ Validation probabilities generated")

# ============================================================
# THRESHOLD TUNING
# ============================================================

print("\n========================================")
print("TUNING LABEL THRESHOLDS")
print("========================================")

thresholds = {}

threshold_values = np.arange(
    0.10,
    0.91,
    0.01
)

threshold_results = []

for label_number, label in enumerate(issue_flag_columns):

    y_true = Y_val[:, label_number]
    probabilities = val_probabilities[:, label_number]

    best_threshold = 0.50
    best_f1 = -1.0

    for threshold in threshold_values:

        y_pred = (
            probabilities >= threshold
        ).astype(int)

        score = f1_score(
            y_true,
            y_pred,
            zero_division=0
        )

        if score > best_f1:

            best_f1 = score
            best_threshold = float(threshold)

    thresholds[label] = best_threshold

    threshold_results.append({
        "label": label,
        "best_threshold": best_threshold,
        "validation_f1": best_f1
    })

    print(
        f"{label}: "
        f"threshold={best_threshold:.2f}, "
        f"validation F1={best_f1:.4f}"
    )

threshold_df = pd.DataFrame(threshold_results)

print("\n✅ Threshold tuning complete")

display(threshold_df)

# ============================================================
# VALIDATION PREDICTIONS USING TUNED THRESHOLDS
# ============================================================

val_pred = np.zeros_like(
    Y_val,
    dtype=int
)

for label_number, label in enumerate(issue_flag_columns):

    val_pred[:, label_number] = (
        val_probabilities[:, label_number]
        >= thresholds[label]
    ).astype(int)

# ============================================================
# VALIDATION METRICS
# ============================================================

val_micro_precision = precision_score(
    Y_val,
    val_pred,
    average="micro",
    zero_division=0
)

val_micro_recall = recall_score(
    Y_val,
    val_pred,
    average="micro",
    zero_division=0
)

val_micro_f1 = f1_score(
    Y_val,
    val_pred,
    average="micro",
    zero_division=0
)

val_macro_f1 = f1_score(
    Y_val,
    val_pred,
    average="macro",
    zero_division=0
)

val_hamming = hamming_loss(
    Y_val,
    val_pred
)

val_exact_match = accuracy_score(
    Y_val,
    val_pred
)

print("\n========================================")
print("VALIDATION RESULTS")
print("========================================")

print(f"Micro Precision : {val_micro_precision:.4f}")
print(f"Micro Recall    : {val_micro_recall:.4f}")
print(f"Micro F1        : {val_micro_f1:.4f}")
print(f"Macro F1        : {val_macro_f1:.4f}")
print(f"Hamming Loss    : {val_hamming:.4f}")
print(f"Exact Match     : {val_exact_match:.4f}")

# ============================================================
# FINAL TEST PROBABILITIES
# ============================================================

print("\n========================================")
print("GENERATING FINAL TEST PREDICTIONS")
print("========================================")

test_probabilities = np.zeros(
    (len(test_idx), len(issue_flag_columns)),
    dtype=np.float32
)

for label_number, label in enumerate(issue_flag_columns, start=1):

    model = trained_models[label]

    test_probabilities[:, label_number - 1] = (
        model.predict_proba(X_test)[:, 1]
    )

    remaining = len(issue_flag_columns) - label_number

    print(
        f"Test prediction: "
        f"{label_number}/{len(issue_flag_columns)} | "
        f"Completed={label_number} | "
        f"Remaining={remaining}"
    )

# ============================================================
# APPLY VALIDATION-TUNED THRESHOLDS TO TEST
# ============================================================

test_pred = np.zeros_like(
    Y_test,
    dtype=int
)

for label_number, label in enumerate(issue_flag_columns):

    test_pred[:, label_number] = (
        test_probabilities[:, label_number]
        >= thresholds[label]
    ).astype(int)

# ============================================================
# FINAL TEST METRICS
# ============================================================

test_micro_precision = precision_score(
    Y_test,
    test_pred,
    average="micro",
    zero_division=0
)

test_micro_recall = recall_score(
    Y_test,
    test_pred,
    average="micro",
    zero_division=0
)

test_micro_f1 = f1_score(
    Y_test,
    test_pred,
    average="micro",
    zero_division=0
)

test_macro_precision = precision_score(
    Y_test,
    test_pred,
    average="macro",
    zero_division=0
)

test_macro_recall = recall_score(
    Y_test,
    test_pred,
    average="macro",
    zero_division=0
)

test_macro_f1 = f1_score(
    Y_test,
    test_pred,
    average="macro",
    zero_division=0
)

test_hamming = hamming_loss(
    Y_test,
    test_pred
)

test_exact_match = accuracy_score(
    Y_test,
    test_pred
)

print("\n========================================")
print("🎯 FINAL TEST RESULTS")
print("========================================")

print(f"Micro Precision : {test_micro_precision:.4f}")
print(f"Micro Recall    : {test_micro_recall:.4f}")
print(f"Micro F1        : {test_micro_f1:.4f}")
print(f"Macro Precision : {test_macro_precision:.4f}")
print(f"Macro Recall    : {test_macro_recall:.4f}")
print(f"Macro F1        : {test_macro_f1:.4f}")
print(f"Hamming Loss    : {test_hamming:.4f}")
print(f"Exact Match     : {test_exact_match:.4f}")

# ============================================================
# PER-LABEL CLASSIFICATION REPORT
# ============================================================

print("\n========================================")
print("PER-LABEL TEST REPORT")
print("========================================")

report_dict = classification_report(
    Y_test,
    test_pred,
    target_names=issue_flag_columns,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report_dict).transpose()

display(
    report_df.round(4)
)

# ============================================================
# MULTI-LABEL CONFUSION MATRICES
# ============================================================

print("\n========================================")
print("CONFUSION MATRICES")
print("========================================")

confusion_matrices = multilabel_confusion_matrix(
    Y_test,
    test_pred
)

for i, label in enumerate(issue_flag_columns):

    cm = confusion_matrices[i]

    print(f"\n--- {label} ---")

    cm_df = pd.DataFrame(
        cm,
        index=[
            "Actual 0",
            "Actual 1"
        ],
        columns=[
            "Predicted 0",
            "Predicted 1"
        ]
    )

    display(cm_df)

# ============================================================
# SAVE METRICS
# ============================================================

MODEL_DIR = os.path.join(
    DATA_DIR,
    "trained_models"
)

metrics = {
    "training_records": int(len(train_idx)),
    "validation_records": int(len(val_idx)),
    "test_records": int(len(test_idx)),

    "embedding_model":
        "sentence-transformers/all-MiniLM-L6-v2",

    "embedding_dimension": 384,

    "validation": {
        "micro_precision": float(val_micro_precision),
        "micro_recall": float(val_micro_recall),
        "micro_f1": float(val_micro_f1),
        "macro_f1": float(val_macro_f1),
        "hamming_loss": float(val_hamming),
        "exact_match": float(val_exact_match)
    },

    "test": {
        "micro_precision": float(test_micro_precision),
        "micro_recall": float(test_micro_recall),
        "micro_f1": float(test_micro_f1),
        "macro_precision": float(test_macro_precision),
        "macro_recall": float(test_macro_recall),
        "macro_f1": float(test_macro_f1),
        "hamming_loss": float(test_hamming),
        "exact_match": float(test_exact_match)
    },

    "thresholds": thresholds
}

metrics_path = os.path.join(
    MODEL_DIR,
    "multilabel_evaluation_metrics.json"
)

with open(metrics_path, "w") as f:
    json.dump(
        metrics,
        f,
        indent=2
    )

# Save threshold table
threshold_path = os.path.join(
    MODEL_DIR,
    "multilabel_thresholds.csv"
)

threshold_df.to_csv(
    threshold_path,
    index=False
)

# Save classification report
report_path = os.path.join(
    MODEL_DIR,
    "multilabel_classification_report.csv"
)

report_df.to_csv(
    report_path
)

# Save test predictions
test_predictions_df = complaints_df.iloc[
    test_idx
][[
    "complaint_id",
    "issue_id",
    "complaint_text"
]].copy()

for i, label in enumerate(issue_flag_columns):

    test_predictions_df[
        f"actual_{label}"
    ] = Y_test[:, i]

    test_predictions_df[
        f"predicted_{label}"
    ] = test_pred[:, i]

    test_predictions_df[
        f"probability_{label}"
    ] = test_probabilities[:, i]

test_predictions_path = os.path.join(
    MODEL_DIR,
    "multilabel_test_predictions.csv"
)

test_predictions_df.to_csv(
    test_predictions_path,
    index=False
)

# ============================================================
# FINAL VERIFICATION
# ============================================================

print("\n========================================")
print("🎉 STEP 9 COMPLETE")
print("========================================")

print("Test records evaluated:", len(test_idx))
print("Remaining test records:", 0)

print("\nSaved evaluation files:")

print("\n1. Metrics:")
print(metrics_path)

print("\n2. Thresholds:")
print(threshold_path)

print("\n3. Classification report:")
print(report_path)

print("\n4. Test predictions:")
print(test_predictions_path)

print("\n✅ Final test evaluation completed")
print("✅ Test set was evaluated after threshold tuning")
print("✅ Results saved to Google Drive")

print("\nNext: Step 10/12 — Prepare the 16,000 relationship-pair dataset")
```

```python
# @title 🧩 Step 10C/12 — Corrected Leakage-Safe Relationship Features + Pair-Group Split

import os
import json
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit

print("===== GrievanceIQ Training Progress =====")
print("Step: 10 / 12")
print("Part C: Corrected relationship feature engineering")
print()

# ============================================================
# 1. LOAD PREPARED DATA
# ============================================================

prepared_path = os.path.join(
    DATA_DIR,
    "relationship_pairs_16000_prepared.csv"
)

relationships_prepared = pd.read_csv(
    prepared_path
)

print("Prepared dataset shape:", relationships_prepared.shape)

assert len(relationships_prepared) == 16000

# ============================================================
# 2. CHECK RELATIONSHIP CLASSES
# ============================================================

print("\n===== RELATIONSHIP CLASSES =====")

display(
    relationships_prepared["relationship_label"]
    .value_counts()
)

# ============================================================
# 3. CATEGORY MAPPING
# ============================================================

FINE_TO_COARSE = {
    "Road Damage": "Roads",
    "Roadside Flooding": "Roads",
    "Water Leakage": "Water Supply",
    "Electric Pole": "Electricity",
    "Drainage": "Drainage",
    "Garbage": "Waste Management",
    "Streetlight": "Public Infrastructure",
    "Traffic Signal": "Public Infrastructure",
    "Tree Hazard": "Public Infrastructure"
}

COARSE_CATEGORIES = [
    "Roads",
    "Water Supply",
    "Electricity",
    "Drainage",
    "Waste Management",
    "Public Infrastructure",
    "Other"
]

def parse_categories(value):

    if pd.isna(value):
        return []

    return [
        x.strip()
        for x in str(value).split("|")
        if x.strip()
    ]


def convert_to_coarse_categories(value):

    fine_categories = parse_categories(value)

    coarse_categories = []

    for category in fine_categories:

        mapped = FINE_TO_COARSE.get(
            category,
            "Other"
        )

        if mapped not in coarse_categories:
            coarse_categories.append(mapped)

    if not coarse_categories:
        coarse_categories = ["Other"]

    return coarse_categories


print("\nConverting fine-grained categories...")

relationships_prepared["coarse_categories_a"] = (
    relationships_prepared["category_a"]
    .apply(convert_to_coarse_categories)
)

relationships_prepared["coarse_categories_b"] = (
    relationships_prepared["category_b"]
    .apply(convert_to_coarse_categories)
)

relationships_prepared["coarse_category_a"] = (
    relationships_prepared["coarse_categories_a"]
    .apply(lambda x: x[0])
)

relationships_prepared["coarse_category_b"] = (
    relationships_prepared["coarse_categories_b"]
    .apply(lambda x: x[0])
)

relationships_prepared["same_category"] = (
    relationships_prepared["coarse_category_a"]
    ==
    relationships_prepared["coarse_category_b"]
).astype(int)

# ============================================================
# 4. LOCATION SCORE
# ============================================================

distance_km = pd.to_numeric(
    relationships_prepared["distance_km"],
    errors="coerce"
).fillna(999999)

relationships_prepared["location_score"] = (
    1.0
    -
    np.clip(
        distance_km / 0.5,
        0.0,
        1.0
    )
)

# ============================================================
# 5. TEMPORAL SCORE
# ============================================================

time_difference = pd.to_numeric(
    relationships_prepared["time_difference_hours"],
    errors="coerce"
).fillna(999999)

relationships_prepared["temporal_score"] = (
    1.0
    -
    np.clip(
        time_difference / 48.0,
        0.0,
        1.0
    )
)

# ============================================================
# 6. DUPLICATE SCORE
# ============================================================

relationships_prepared["duplicate_score"] = (
    0.50
    * relationships_prepared["semantic_similarity"]
    +
    0.30
    * relationships_prepared["location_score"]
    +
    0.20
    * relationships_prepared["temporal_score"]
)

relationships_prepared["duplicate_flag"] = (
    (
        relationships_prepared["duplicate_score"] >= 0.80
    )
    &
    (
        relationships_prepared["semantic_similarity"] >= 0.75
    )
).astype(int)

# ============================================================
# 7. BUILD EXACTLY 20 BACKEND-COMPATIBLE FEATURES
# ============================================================

feature_columns = [
    "semantic_similarity",
    "location_score",
    "temporal_score",
    "duplicate_score",
    "duplicate_flag",
    "same_category"
]

# 7 category A one-hot features
for category in COARSE_CATEGORIES:

    column = (
        "category_a_"
        + category.lower().replace(" ", "_")
    )

    relationships_prepared[column] = (
        relationships_prepared["coarse_category_a"]
        == category
    ).astype(int)

    feature_columns.append(column)

# 7 category B one-hot features
for category in COARSE_CATEGORIES:

    column = (
        "category_b_"
        + category.lower().replace(" ", "_")
    )

    relationships_prepared[column] = (
        relationships_prepared["coarse_category_b"]
        == category
    ).astype(int)

    feature_columns.append(column)

# ============================================================
# 8. VERIFY FEATURE COUNT
# ============================================================

print("\n===== FEATURE COUNT =====")

print("Feature count:", len(feature_columns))

assert len(feature_columns) == 20

print("✅ Exactly 20 features")

print("\nFeatures:")

for number, feature in enumerate(
    feature_columns,
    start=1
):
    print(
        f"{number:02d}. {feature}"
    )

# ============================================================
# 9. LEAKAGE CHECK
# ============================================================

print("\n===== LEAKAGE CHECK =====")

for forbidden in [
    "same_issue",
    "relationship_label"
]:

    if forbidden in feature_columns:
        raise ValueError(
            f"LEAKAGE DETECTED: {forbidden}"
        )

    print(
        f"{forbidden}: NOT USED ✅"
    )

if "category_overlap_binary" in feature_columns:

    raise ValueError(
        "category_overlap_binary should not be present"
    )

print(
    "category_overlap_binary: NOT USED ✅"
)

# ============================================================
# 10. CREATE X AND y
# ============================================================

X_relationship = (
    relationships_prepared[
        feature_columns
    ]
    .astype(np.float32)
    .values
)

y_relationship = (
    relationships_prepared[
        "relationship_label"
    ]
    .astype(str)
    .values
)

print("\n===== MATRIX SHAPES =====")

print(
    "X shape:",
    X_relationship.shape
)

print(
    "y shape:",
    y_relationship.shape
)

assert X_relationship.shape == (
    16000,
    20
)

assert y_relationship.shape == (
    16000,
)

print("✅ Feature matrix correct")

# ============================================================
# 11. NUMERIC CHECK
# ============================================================

print("\n===== NUMERIC CHECK =====")

nan_count = np.isnan(
    X_relationship
).sum()

inf_count = np.isinf(
    X_relationship
).sum()

print("NaN:", nan_count)
print("Infinite:", inf_count)

assert nan_count == 0
assert inf_count == 0

print("✅ No NaN")
print("✅ No infinite values")

# ============================================================
# 12. CANONICAL ISSUE-PAIR GROUP
# ============================================================
#
# Example:
# I0001 + I0042
# I0042 + I0001
#
# become the same group.
#
# This avoids exact pair-group leakage.
# ============================================================

print("\n========================================")
print("BUILDING CANONICAL ISSUE-PAIR GROUPS")
print("========================================")

issue_a = (
    relationships_prepared[
        "issue_id_a"
    ]
    .astype(str)
)

issue_b = (
    relationships_prepared[
        "issue_id_b"
    ]
    .astype(str)
)

relationships_prepared[
    "pair_group"
] = [
    "|".join(
        sorted([a, b])
    )
    for a, b in zip(
        issue_a,
        issue_b
    )
]

unique_pair_groups = (
    relationships_prepared[
        "pair_group"
    ]
    .nunique()
)

print(
    "Unique issue-pair groups:",
    unique_pair_groups
)

print(
    "Total relationship rows:",
    len(relationships_prepared)
)

# ============================================================
# 13. GROUPED TRAIN / VALIDATION / TEST SPLIT
# ============================================================

print("\n========================================")
print("PAIR-GROUP TRAIN / VALIDATION / TEST SPLIT")
print("========================================")

groups = (
    relationships_prepared[
        "pair_group"
    ]
    .astype(str)
    .values
)

# 85% train+validation
# 15% test

gss_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=42
)

train_val_idx, test_idx_rel = next(
    gss_test.split(
        X_relationship,
        y_relationship,
        groups=groups
    )
)

# From 85%:
# ~70% total train
# ~15% total validation

gss_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.17647058823529413,
    random_state=42
)

train_rel_idx, val_rel_idx = next(
    gss_val.split(
        X_relationship[train_val_idx],
        y_relationship[train_val_idx],
        groups=groups[train_val_idx]
    )
)

relationship_train_idx = (
    train_val_idx[
        train_rel_idx
    ]
)

relationship_val_idx = (
    train_val_idx[
        val_rel_idx
    ]
)

relationship_test_idx = (
    test_idx_rel
)

# ============================================================
# 14. COUNTS
# ============================================================

total_pairs = len(
    relationships_prepared
)

train_count = len(
    relationship_train_idx
)

val_count = len(
    relationship_val_idx
)

test_count = len(
    relationship_test_idx
)

accounted = (
    train_count
    +
    val_count
    +
    test_count
)

remaining = (
    total_pairs
    -
    accounted
)

print("\n===== RELATIONSHIP SPLIT COUNTS =====")

print(
    "Total pairs:      ",
    total_pairs
)

print(
    "Training pairs:   ",
    train_count
)

print(
    "Validation pairs: ",
    val_count
)

print(
    "Test pairs:       ",
    test_count
)

print(
    "Remaining pairs:  ",
    remaining
)

print("\n===== PERCENTAGES =====")

print(
    f"Training:   "
    f"{train_count / total_pairs * 100:.2f}%"
)

print(
    f"Validation: "
    f"{val_count / total_pairs * 100:.2f}%"
)

print(
    f"Test:       "
    f"{test_count / total_pairs * 100:.2f}%"
)

assert accounted == total_pairs
assert remaining == 0

print("\n✅ All 16,000 relationship pairs accounted for")

# ============================================================
# 15. CHECK PAIR-GROUP LEAKAGE
# ============================================================

train_groups = set(
    groups[
        relationship_train_idx
    ]
)

val_groups = set(
    groups[
        relationship_val_idx
    ]
)

test_groups = set(
    groups[
        relationship_test_idx
    ]
)

train_val_overlap = (
    train_groups
    &
    val_groups
)

train_test_overlap = (
    train_groups
    &
    test_groups
)

val_test_overlap = (
    val_groups
    &
    test_groups
)

print("\n===== PAIR-GROUP LEAKAGE CHECK =====")

print(
    "Train ↔ Validation:",
    len(train_val_overlap)
)

print(
    "Train ↔ Test:",
    len(train_test_overlap)
)

print(
    "Validation ↔ Test:",
    len(val_test_overlap)
)

assert len(train_val_overlap) == 0
assert len(train_test_overlap) == 0
assert len(val_test_overlap) == 0

print(
    "\n✅ No exact issue-pair group leakage"
)

# ============================================================
# 16. LABEL DISTRIBUTION
# ============================================================

print("\n===== LABEL DISTRIBUTION =====")

train_labels = (
    relationships_prepared
    .iloc[relationship_train_idx]
    ["relationship_label"]
    .value_counts()
)

val_labels = (
    relationships_prepared
    .iloc[relationship_val_idx]
    ["relationship_label"]
    .value_counts()
)

test_labels = (
    relationships_prepared
    .iloc[relationship_test_idx]
    ["relationship_label"]
    .value_counts()
)

distribution_df = pd.DataFrame({

    "Train": train_labels,

    "Validation": val_labels,

    "Test": test_labels

}).fillna(0).astype(int)

display(
    distribution_df
)

# ============================================================
# 17. SAVE FEATURE DATASET
# ============================================================

features_path = os.path.join(
    DATA_DIR,
    "relationship_training_features_16000.csv"
)

relationships_prepared.to_csv(
    features_path,
    index=False
)

print("\n✅ Feature dataset saved:")
print(features_path)

# ============================================================
# 18. SAVE SPLIT INDICES
# ============================================================

split_indices_path = os.path.join(
    DATA_DIR,
    "relationship_split_indices.npz"
)

np.savez(
    split_indices_path,

    train_idx=relationship_train_idx,

    val_idx=relationship_val_idx,

    test_idx=relationship_test_idx
)

print("\n✅ Split indices saved:")
print(split_indices_path)

# ============================================================
# 19. SAVE SPLIT INFORMATION
# ============================================================

split_info = {

    "total_pairs": int(total_pairs),

    "train_pairs": int(train_count),

    "validation_pairs": int(val_count),

    "test_pairs": int(test_count),

    "remaining_pairs": int(remaining),

    "feature_count": int(len(feature_columns)),

    "features": feature_columns,

    "grouping_method":
        "canonical_issue_pair",

    "random_state": 42,

    "same_issue_used_as_feature": False,

    "relationship_label_used_as_feature": False,

    "category_overlap_binary_used_as_feature": False

}

split_info_path = os.path.join(
    DATA_DIR,
    "relationship_split_info.json"
)

with open(
    split_info_path,
    "w"
) as f:

    json.dump(
        split_info,
        f,
        indent=2
    )

print("\n✅ Split information saved:")
print(split_info_path)

# ============================================================
# FINAL STATUS
# ============================================================

print("\n========================================")
print("🎉 STEP 10C CORRECTED COMPLETE")
print("========================================")

print(
    "Features:",
    len(feature_columns)
)

print(
    "Total pairs:",
    total_pairs
)

print(
    "Training:",
    train_count
)

print(
    "Validation:",
    val_count
)

print(
    "Test:",
    test_count
)

print(
    "Remaining:",
    remaining
)

print("\n✅ 20 backend-compatible features")
print("✅ Leakage fields excluded")
print("✅ Pair groups separated")
print("✅ All 16,000 pairs accounted for")

print("\nNext:")
print("Step 11/12 — Train Duplicate / Similar / Related / Independent classifier")
```

```python
# @title 🌲 Step 11C — FAST Backend-Compatible Relationship Random Forest

import os
import json
import subprocess
import textwrap

# ============================================================
# CONFIGURATION
# ============================================================

MODEL_DIR = os.path.join(
    DATA_DIR,
    "trained_models"
)

payload_path = os.path.join(
    MODEL_DIR,
    "relationship_training_payload.json"
)

node_script_path = (
    "/content/"
    "train_grievanceiq_relationship_model_fast.js"
)

model_output_path = os.path.join(
    MODEL_DIR,
    "relationship_model.json"
)

metadata_output_path = os.path.join(
    MODEL_DIR,
    "relationship_model_metadata.json"
)

predictions_output_path = os.path.join(
    MODEL_DIR,
    "relationship_predictions.json"
)

# ============================================================
# FAST MODEL SETTINGS
# ============================================================

N_ESTIMATORS = 30
MAX_FEATURES = 0.5
SEED = 42
REPLACEMENT = True
USE_SAMPLE_BAGGING = True

# ============================================================
# NODE.JS TRAINING SCRIPT
# ============================================================

script = r'''
const fs = require("fs");

const {
    RandomForestClassifier
} = require("ml-random-forest");

const payloadPath = process.argv[2];
const modelPath = process.argv[3];
const metadataPath = process.argv[4];
const predictionsPath = process.argv[5];

const N_ESTIMATORS = 30;
const MAX_FEATURES = 0.5;
const SEED = 42;
const REPLACEMENT = true;
const USE_SAMPLE_BAGGING = true;

const LABELS = [
    "Duplicate",
    "Similar",
    "Related",
    "Independent"
];

const payload = JSON.parse(
    fs.readFileSync(
        payloadPath,
        "utf8"
    )
);

const features = payload.features;

const XTrain = payload.X_train;
const yTrain = payload.y_train;

const XVal = payload.X_val;
const yVal = payload.y_val;

const XTest = payload.X_test;
const yTest = payload.y_test;

function evaluate(yTrue, yPred, numClasses) {

    const confusion = Array.from(
        { length: numClasses },
        () => new Array(numClasses).fill(0)
    );

    for (
        let i = 0;
        i < yTrue.length;
        i++
    ) {
        confusion[
            yTrue[i]
        ][
            yPred[i]
        ]++;
    }

    let correct = 0;

    for (
        let c = 0;
        c < numClasses;
        c++
    ) {
        correct += confusion[c][c];
    }

    const accuracy =
        correct / yTrue.length;

    const perClass = [];

    for (
        let c = 0;
        c < numClasses;
        c++
    ) {

        const tp =
            confusion[c][c];

        let fp = 0;
        let fn = 0;

        for (
            let j = 0;
            j < numClasses;
            j++
        ) {

            if (j !== c) {

                fp +=
                    confusion[j][c];

                fn +=
                    confusion[c][j];
            }
        }

        const precision =
            (tp + fp) > 0
                ? tp / (tp + fp)
                : 0;

        const recall =
            (tp + fn) > 0
                ? tp / (tp + fn)
                : 0;

        const f1 =
            (precision + recall) > 0
                ? (
                    2 *
                    precision *
                    recall
                  ) /
                  (
                    precision +
                    recall
                  )
                : 0;

        perClass.push({

            label:
                LABELS[c],

            precision:
                Number(
                    precision.toFixed(4)
                ),

            recall:
                Number(
                    recall.toFixed(4)
                ),

            f1:
                Number(
                    f1.toFixed(4)
                ),

            support:
                tp + fn
        });
    }

    return {

        accuracy:
            Number(
                accuracy.toFixed(4)
            ),

        confusionMatrix:
            confusion,

        perClass
    };
}

console.log(
    "========================================"
);

console.log(
    "GrievanceIQ Relationship Random Forest"
);

console.log(
    "FAST BACKEND-COMPATIBLE VERSION"
);

console.log(
    "========================================"
);

console.log("");

console.log(
    "Training pairs:",
    XTrain.length
);

console.log(
    "Validation pairs:",
    XVal.length
);

console.log(
    "Test pairs:",
    XTest.length
);

console.log(
    "Features:",
    features.length
);

console.log(
    "Classes:",
    LABELS.length
);

console.log("");

console.log(
    "===== MODEL CONFIGURATION ====="
);

console.log(
    "Trees:",
    N_ESTIMATORS
);

console.log(
    "Max features:",
    MAX_FEATURES
);

console.log(
    "Replacement:",
    REPLACEMENT
);

console.log(
    "Sample bagging:",
    USE_SAMPLE_BAGGING
);

console.log(
    "Seed:",
    SEED
);

console.log("");

console.log(
    "===== TRAINING ====="
);

console.log(
    "Progress:"
);

console.log(
    "1/4 Dataset loaded ✅"
);

console.log(
    "2/4 20-feature matrix ready ✅"
);

console.log(
    "3/4 Training Random Forest..."
);

const classifier =
    new RandomForestClassifier({

        nEstimators:
            N_ESTIMATORS,

        maxFeatures:
            MAX_FEATURES,

        replacement:
            REPLACEMENT,

        seed:
            SEED,

        useSampleBagging:
            USE_SAMPLE_BAGGING
    });

const startTime =
    Date.now();

classifier.train(
    XTrain,
    yTrain
);

const trainingSeconds =
    (
        Date.now()
        -
        startTime
    ) / 1000;

console.log(
    "4/4 Training complete ✅"
);

console.log(
    "Training time:",
    trainingSeconds.toFixed(2),
    "seconds"
);

console.log("");

console.log(
    "===== VALIDATION ====="
);

const yValPred =
    classifier.predict(
        XVal
    );

const validationMetrics =
    evaluate(
        yVal,
        yValPred,
        LABELS.length
    );

console.log(
    "Validation predictions:",
    yValPred.length,
    "/",
    yVal.length
);

console.log(
    "Validation accuracy:",
    validationMetrics.accuracy
);

for (
    const result
    of validationMetrics.perClass
) {

    console.log(
        result.label,
        "F1=",
        result.f1
    );
}

console.log("");

console.log(
    "===== FINAL TEST ====="
);

const yTestPred =
    classifier.predict(
        XTest
    );

const testMetrics =
    evaluate(
        yTest,
        yTestPred,
        LABELS.length
    );

console.log(
    "Test predictions:",
    yTestPred.length,
    "/",
    yTest.length
);

console.log(
    "Test accuracy:",
    testMetrics.accuracy
);

console.log("");

console.log(
    "===== TEST PER-CLASS RESULTS ====="
);

for (
    const result
    of testMetrics.perClass
) {

    console.log(
        result.label.padEnd(15),
        "Precision=",
        result.precision.toFixed(4),
        "Recall=",
        result.recall.toFixed(4),
        "F1=",
        result.f1.toFixed(4),
        "Support=",
        result.support
    );
}

console.log("");

console.log(
    "===== TEST CONFUSION MATRIX ====="
);

console.log(
    "Rows = Actual"
);

console.log(
    "Columns = Predicted"
);

console.log(
    "             ",
    LABELS.join(" | ")
);

for (
    let i = 0;
    i < testMetrics.confusionMatrix.length;
    i++
) {

    console.log(
        LABELS[i].padEnd(13),
        testMetrics.confusionMatrix[i]
            .join(" | ")
    );
}

console.log("");

console.log(
    "===== SAVING MODEL ====="
);

const modelJson =
    classifier.toJSON();

fs.writeFileSync(
    modelPath,
    JSON.stringify(
        modelJson
    ),
    "utf8"
);

console.log(
    "✅ Model saved:",
    modelPath
);

const metadata = {

    version:
        "2.0.0-fast",

    phase:
        3,

    trainingDate:
        new Date().toISOString(),

    datasetSize:
        XTrain.length +
        XVal.length +
        XTest.length,

    trainSize:
        XTrain.length,

    validationSize:
        XVal.length,

    testSize:
        XTest.length,

    featureCount:
        features.length,

    features:

        features,

    labels:

        LABELS,

    hyperparameters: {

        nEstimators:
            N_ESTIMATORS,

        maxFeatures:
            MAX_FEATURES,

        replacement:
            REPLACEMENT,

        useSampleBagging:
            USE_SAMPLE_BAGGING,

        seed:
            SEED
    },

    splitMethod:
        "canonical_issue_pair_group",

    leakageControls: {

        same_issue_used:
            false,

        relationship_label_used:
            false,

        pair_group_overlap:
            false
    },

    validationMetrics:
        validationMetrics,

    testMetrics:
        testMetrics,

    trainingTimeSeconds:
        trainingSeconds,

    notes:
        "Synthetic held-out evaluation. "
        +
        "Does not establish real-world generalization."
};

fs.writeFileSync(
    metadataPath,
    JSON.stringify(
        metadata,
        null,
        2
    ),
    "utf8"
);

console.log(
    "✅ Metadata saved:",
    metadataPath
);

const predictions = {

    labels:

        LABELS,

    validation: {

        y_true:
            yVal,

        y_pred:
            yValPred,

        metrics:
            validationMetrics
    },

    test: {

        y_true:
            yTest,

        y_pred:
            yTestPred,

        metrics:
            testMetrics
    }
};

fs.writeFileSync(
    predictionsPath,
    JSON.stringify(
        predictions,
        null,
        2
    ),
    "utf8"
);

console.log(
    "✅ Predictions saved:",
    predictionsPath
);

console.log("");

console.log(
    "========================================"
);

console.log(
    "🎉 STEP 11C COMPLETE"
);

console.log(
    "========================================"
);

console.log(
    "Training:",
    XTrain.length,
    "/",
    XTrain.length
);

console.log(
    "Validation:",
    XVal.length,
    "/",
    XVal.length
);

console.log(
    "Test:",
    XTest.length,
    "/",
    XTest.length
);

console.log(
    "Remaining:",
    0
);
'''

# ============================================================
# WRITE SCRIPT
# ============================================================

with open(
    node_script_path,
    "w"
) as f:

    f.write(script)

print(
    "✅ New fast Node training script created:"
)

print(
    node_script_path
)

print("")

print(
    "Starting FAST backend-compatible training..."
)

print(
    "Expected: a few minutes rather than the previous long run."
)

print("")

# ============================================================
# RUN WITHOUT CAPTURING OUTPUT
# ============================================================
# This lets Node's output appear immediately in Colab.

result = subprocess.run(
    [
        "node",
        node_script_path,
        payload_path,
        model_output_path,
        metadata_output_path,
        predictions_output_path
    ]
)

# ============================================================
# CHECK RESULT
# ============================================================

if result.returncode != 0:

    raise RuntimeError(
        f"Node training failed with exit code "
        f"{result.returncode}"
    )

# ============================================================
# CHECK FILES
# ============================================================

print("")
print(
    "===== FINAL FILE CHECK ====="
)

for file_path in [

    model_output_path,

    metadata_output_path,

    predictions_output_path

]:

    if not os.path.exists(
        file_path
    ):

        raise FileNotFoundError(
            file_path
        )

    size_kb = (
        os.path.getsize(
            file_path
        )
        / 1024
    )

    print(
        f"✅ {os.path.basename(file_path)} "
        f"({size_kb:.1f} KB)"
    )

print("")
print(
    "🎉 STEP 11C SUCCESSFULLY COMPLETED"
)

print(
    "Training pairs: 11134 / 11134"
)

print(
    "Validation pairs: 2446 / 2446"
)

print(
    "Test pairs: 2420 / 2420"
)

print(
    "Remaining: 0"
)
```
